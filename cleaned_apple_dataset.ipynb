```python
Q1 = df ['launch_price_usd'] .quantile (0.25)
Q3 = df ['launch_price_usd'] .quantile (0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR 
upper = Q3 +1.5 * IQR
outlier =  df [( df['launch_price_usd'] < lower) | (df['launch_price_usd']> upper)]
print ( f"{len(outlier)} outlier found")
outlier['product_category'].value_counts()
```

```python
Q1 = df ['current_price_usd'] .quantile (0.25)
Q3 = df ['current_price_usd'] .quantile (0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR 
upper = Q3 +1.5 * IQR
outlier =  df [( df['current_price_usd'] < lower) | (df['current_price_usd']> upper)]
print ( f"{len(outlier)} outlier found")
outlier['product_category'].value_counts()
```

```python
Q1 = df ['discount_pct'] .quantile (0.25)
Q3 = df ['discount_pct'] .quantile (0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR 
upper = Q3 +1.5 * IQR
outlier =  df [( df['discount_pct'] < lower) | (df['discount_pct']> upper)]
print ( f"{len(outlier)} outlier found")
outlier['product_category'].value_counts()
```

```python
mean = df['current_price_inr'].mean()
std = df['current_price_inr'].std()
df['z_score'] = (df['current_price_inr'] -mean) / std 
outlier_z = df [df['z_score'].abs()>3]
outlier_z
```
